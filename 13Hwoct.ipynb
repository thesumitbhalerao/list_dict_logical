#### Task 01 : -> WAP to get a unique element list from above list without changing data position.

```python
l=[5,4,5,4,2,1,3,1,6,4,2,3,5,7,6]
unique_element=[]
for data in l:
    if data not in unique_element:
        unique_element.append(data)
print(unique_element)
```

#### Task 02 : -> find the freq. of every element from above input list in dictionary.

```python
l=[5,4,5,4,2,1,3,1,6,4,2,3,5,7,6]
freq_ele={}
for i in l:
    if i in freq_ele:
        freq_ele[i] = freq_ele[i]+1
    else:
        freq_ele[i]=1
print(freq_ele)
```
